```python
from napari import Viewer

import sys
from faim_ipa.utils import get_git_root
from pathlib import Path
from ipyfilechooser import FileChooser
from ipywidgets import widgets
from IPython.display import display

sys.path.append(str(get_git_root()))

from source.s01_convert_to_zarr.create_selection_csv_utils import (
    get_experiment_widget,
    load_positions,
)
from source.s03_measure.MoltAnnotator import MoltAnnotator

viewer = Viewer()
```

# Choose the git-repo on the file-server

```python
fc = FileChooser(
    path=get_git_root(),
    layout=widgets.Layout(width="100%"),
    show_only_dirs=True,
    title="Select Processing Directory",
)
```

```python
display(fc)
```

```python
root_dir = Path(fc.selected)
```

# Select the experiment 

```python
exp = get_experiment_widget(root_dir)
```

```python
display(exp)
```

```python
config, position_paths = load_positions(exp, root_dir, from_input_dir=False)
```

```python
annot = MoltAnnotator(
    viewer, position_paths, brightfield_channel_index=0, exp_path=exp.value
)
```

```python
viewer.bind_key("e", lambda _: annot.set_m1(), overwrite=True)
viewer.bind_key("r", lambda _: annot.set_m2(), overwrite=True)
viewer.bind_key("d", lambda _: annot.set_m3(), overwrite=True)
viewer.bind_key("f", lambda _: annot.set_m4(), overwrite=True)
viewer.bind_key("w", lambda _: annot.add_next(), overwrite=True)
viewer.bind_key("q", lambda _: annot.add_previous(), overwrite=True)
```

# Annotate all positions in napari

## Hotkeys
* `e`: Set current time-point as 1st molting point. Pressing a 2nd time on the same timepoint will reset to `nan`.
* `r`: Set current time-point as 2nd molting point. Pressing a 2nd time on the same timepoint will reset to `nan`.
* `d`: Set current time-point as 3rd molting point. Pressing a 2nd time on the same timepoint will reset to `nan`.
* `f`: Set current time-point as 4th molting point. Pressing a 2nd time on the same timepoint will reset to `nan`.
* `w`: Load next position.
* `q`: Load previous position.

```python
molt_df = annot.get_molt_df()
```

```python
molt_df
```

```python
molt_csv_path = exp.value / "molt.csv"
molt_df.to_csv(molt_csv_path, index=False)

print(f"Molt annotations saved to: {str(molt_csv_path)}")
```
