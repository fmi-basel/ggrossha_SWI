```python
from napari import Viewer

import sys
from faim_ipa.utils import get_git_root
from pathlib import Path
from ipyfilechooser import FileChooser
from ipywidgets import widgets
from IPython.display import display

sys.path.append(str(get_git_root()))

from source.s01_convert_to_zarr.create_selection_csv_utils import (
    get_experiment_widget,
    load_positions,
    Annotator,
)

viewer = Viewer()
```

# Choose the git-repo on the file-server

```python
fc = FileChooser(
    path=get_git_root(),
    layout=widgets.Layout(width="100%"),
    show_only_dirs=True,
    title="Select Processing Directory",
)
display(fc)
```

```python
root_dir = Path(fc.selected)
```

# Select the experiment 

```python
exp = get_experiment_widget(root_dir)
display(exp)
```

```python
config, position_paths = load_positions(exp, root_dir)
```

```python
annot = Annotator(viewer, position_paths)
```

```python
viewer.bind_key("e", lambda _: annot.set_start_time(), overwrite=True)
viewer.bind_key("r", lambda _: annot.set_end_time(), overwrite=True)
viewer.bind_key("w", lambda _: annot.add_next(), overwrite=True)
viewer.bind_key("q", lambda _: annot.add_previous(), overwrite=True)
viewer.bind_key("s", lambda _: annot.skip(), overwrite=True)
```

# Annotate all positions in napari

## Hotkeys
* `e`: Set current time-point as starting point (e.g. hatch).
* `r`: Set current time-point as end point (e.g. escape).
* `w`: Load next position.
* `q`: Load previous position.
* `s`: Skip the current position.

```python
selection_df = annot.get_selection_df()
```

```python
selection_csv_path = exp.value / "selection.csv"
selection_df.to_csv(selection_csv_path, index=False)

config.preview = False
config.save(config_file=exp.value / config.config_name())

print(f"Selection saved to: {str(selection_csv_path)}")
```
